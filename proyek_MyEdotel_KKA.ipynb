```python
from abc import ABC, abstractmethod

# Helper function untuk format Rupiah
def format_rp(angka):
    return f"Rp {int(angka):,.0f}".replace(",", ".")

# 1. Abstraction & Encapsulation
class KamarHotel(ABC):
    def __init__(self, nama_kamar, harga_dasar, stok=0):
        self.nama_kamar = nama_kamar
        self.__harga_dasar = harga_dasar
        self.__stok = stok

    @property
    def stok(self):
        return self.__stok

    @property
    def harga_dasar(self):
        return self.__harga_dasar

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.jenis_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.jenis_kamar}: {self.__stok} unit.")

    @abstractmethod
    def tampilkan_detail(self, jumlah_malam, nomor):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

# 2. Inheritance & Polymorphism (Kamar Deluxe)
class KamarDeluxe(KamarHotel):
    jenis_kamar = "Kamar Deluxe"

    def __init__(self, nama_kamar, harga_dasar, fasilitas, stok=0):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.fasilitas = fasilitas
        self.pajak = 0.10

    def hitung_harga_total(self, jumlah_malam):
        total_per_malam = self.harga_dasar + (self.harga_dasar * self.pajak)
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam, nomor):
        pajak_rp = self.harga_dasar * self.pajak
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"Harga Dasar/Malam: {format_rp(self.harga_dasar)} | Pajak(10%): {format_rp(pajak_rp)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rp(subtotal)}")

# 3. Inheritance & Polymorphism (Kamar Standard)
class KamarStandard(KamarHotel):
    jenis_kamar = "Kamar Standard"

    def __init__(self, nama_kamar, harga_dasar, kapasitas, stok=0):
        super().__init__(nama_kamar, harga_dasar, stok)
        self.kapasitas = kapasitas
        self.pajak = 0.05

    def hitung_harga_total(self, jumlah_malam):
        total_per_malam = self.harga_dasar + (self.harga_dasar * self.pajak)
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam, nomor):
        pajak_rp = self.harga_dasar * self.pajak
        subtotal = self.hitung_harga_total(jumlah_malam)
        print(f"{nomor}. [STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"Harga Dasar/Malam: {format_rp(self.harga_dasar)} | Pajak(5%): {format_rp(pajak_rp)}")
        print(f"Menginap: {jumlah_malam} malam | Subtotal: {format_rp(subtotal)}")

# Fitur Pemesanan (Di luar class)
def proses_transaksi(daftar_pesanan):
    print("--- STRUK PEMESANAN ---")
    total_tagihan = 0
    for i, pesanan in enumerate(daftar_pesanan, start=1):
        kamar = pesanan['kamar']
        malam = pesanan['malam']

        kamar.tampilkan_detail(malam, i)
        total_tagihan += kamar.hitung_harga_total(malam)

    print("-" * 40)
    print(f"TOTAL TAGIHAN: {format_rp(total_tagihan)}")
    print("-" * 40)

# ==========================================
# Alur Program (User Story)
# ==========================================
if __name__ == "__main__":
    # a) Admin membuat data kamar
    deluxe = KamarDeluxe("Kamar Deluxe Sea View", 1500000, "Private Pool")
    standard = KamarStandard("Kamar Standard Superior", 500000, "2 Orang")

    print("--- SETUP DATA KAMAR ---")
    deluxe.tambah_stok(10)

    # b) Admin mencoba mengisi stok dengan angka negatif
    standard.tambah_stok(-5)

    # Admin mengkoreksi penambahan stok
    standard.tambah_stok(20)

    # c) Tamu memesan kamar
    daftar_pesanan = [
        {'kamar': deluxe, 'malam': 2},
        {'kamar': standard, 'malam': 1}
    ]

    # d) Program menampilkan detail dan tagihan
    proses_transaksi(daftar_pesanan)
```
